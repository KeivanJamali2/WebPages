```python
def round(value: float, decimals: int = 0) -> float:
    """
    Round a value to a specified number of decimal places.

    Args:
        value: The value to round
        decimals: Number of decimal places (default=0)

    Returns:
        Rounded value
    """
    value = str(value)
    i, d = value.split(".")
    if decimals >= len(d):
        return float(value)
    if decimals == 0:
        if int(d[0]) >= 5:
            return float(int(i) + 1)
        else:
            return float(i)
    else:
        temp = d[decimals]
        new_d = d[:decimals]
        if int(temp) >= 5:
            d = int(new_d[-1]) + 1
        else:
            d = new_d
    
    return float(".".join([i, str(new_d)]))
```
